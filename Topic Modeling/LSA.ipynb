```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.pipeline import Pipeline
```

```python
doc1 = "Mango and starberries are the sweet"
doc2 = "Science is her favourite subject"
doc3 = "Mathematics is an important subject"
doc4 = "Lemons and oranges are sour"
doc5 = "Algorithms and databases are tough"
documents = [doc1, doc2, doc3, doc4, doc5]
documents
```

```python
vectorizer = TfidfVectorizer(
    stop_words = 'english', use_idf = True , smooth_idf = True
)
x = vectorizer.fit_transform(documents)
x
```

```python
# n_components = number of topics
lsa = TruncatedSVD(
    n_components = 2 , n_iter = 100 , random_state = 42
)
```

```python
lsa.fit(x)
lsa.components_
```

```python
terms = vectorizer.get_feature_names_out()
len(terms)
```

```python
for index , component in enumerate(lsa.components_):
    termsInComp = zip(terms , component)
    sortedTerms = sorted(termsInComp , key = lambda x: x[1])
    print(f"Concept {index}")
    for term in sortedTerms:
        print(term[0])
    
    print(" ")
```
