### Dealing with Contractions:
---
- Contractions are shortened forms created by combining words.
     - do not -> don't
- Why we should remove (expand) contractions?
    - Removing contractions before forming word vectors helps in dimensionality reduction.
    - Improves semantice clearity.
    - Better negation modeling cause negation words like not,never become separete tokens.
    - Reduce sparsity in BOW and TF-IDF.

```python
import contractions
```

```python
contractions.fix("I'll can't")
```

```python
text = '''I'll be there within 5 min. Shouldn't you be there too? 
          I'd love to see u there my dear. It's awesome to meet new friends.
          We've been waiting for this day for so long.'''
```

```python
expand_words = []
for word in text.split(): 
    expand_words.append(contractions.fix(word))

text = " ".join(expand_words)
text
```

## Coreference resolution.
---
- read from [1](https://neurosys.com/blog/intro-to-coreference-resolution-in-nlp) ,  [2](https://www.netguru.com/glossary/coreference-resolution)
- a process that involves identifying when different words or phrases in a text refer to the same entity or concept.
   Ex: Philip plays the bass because he loves it.
   Here he refers to Philip and it refers to bass.

- It involves identifying when two or more expressions in a text refer to the same entity.
- removing coreference means:Replacing pronouns or indirect mentions with their resolved entities.
- When you SHOULD resolve/remove coreference:
     - Information extraction/knowledge graph
     - Bio or medical documents
     - Ner problems.
- When should not remove:
    - Language modeling/text generation
    - Sentiment Analysis (document-level)
- If your task depends on “who did what to whom” → resolve coreference.
- If your task depends on “overall meaning or tone” → often skip it.

### Doc2Vec Embedding

```python
from gensim.models.doc2vec import Doc2Vec, TaggedDocument
from nltk.tokenize import word_tokenize
```

```python
documents = ["This is the first document",
        "This is the second document",
        "This is the third document",
        "This is the fourth document"]
```

```python
""" 
TaggedDocument takes 2 things: 
   1. A list of tokens more specifically the list of tokens for the current document.
   2. A unique identifier by which this document can be identified.
"""

tagged_docs = [
    TaggedDocument(
        words = word_tokenize(doc.lower()), 
        tags = ["doc_" + str(i)] 
    ) for i , doc in enumerate(documents)
]
tagged_docs
```

```python
# train a doc2vec model 
model = Doc2Vec(vector_size = 50, window = 2, min_count = 1,epochs = 40)
model.build_vocab(tagged_docs)
```

```python
model.train(tagged_docs,total_examples = model.corpus_count, epochs = model.epochs)
```

```python
vector = model.infer_vector(['first' , 'document'])
vector
```
