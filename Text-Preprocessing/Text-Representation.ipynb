```python
import pandas as pd
```

```python
df = pd.DataFrame([
    ['people watch campusX' , 1], 
    ['campusX watch campusX' , 1], 
    ['people write comment' , 0], 
    ['campusX write comment' , 0]
] , columns = ['text' , 'output'])
df
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer()
# Bag of Words
bow = cv.fit_transform(df['text'])
```

```python
# vocab 
cv.vocabulary_
```

```python
bow[0]
```

```python
bow[0].toarray()
```

```python
bow.toarray()
```

```python
cv.transform(["campusX people write comment campusX"]).toarray()
```

```python
cv.transform(["campusX people write comment campusX read"]).toarray()
```

- If any word which is not in our vocab then it will be excluded.
- If binary = True in cv then it replace all values with 1 if value is greater that 1, it is proven that for sentiment analysis, it works good if use binary = True.

### N-gram
---
- Bag of Words is a special N-gram where value of n is (1,1) means uni-gram

```python
bi_gram = CountVectorizer(ngram_range = (2 , 2))
bow = bi_gram.fit_transform(df['text'])
bow.toarray()
```

```python
bi_gram.vocabulary_
```

### TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tf_idf = TfidfVectorizer()
transformed = tf_idf.fit_transform(df['text'])
```

```python
transformed.toarray()
```

```python
tf_idf.idf_
```

```python
print(tf_idf.idf_)
print(tf_idf.get_feature_names_out())
```
