### Lowercasing

```python
def convert_text_into_lowercase(text: str): 
    return text.lower()
```

```python
df['review'] = df['review'].apply(convert_text_into_lowercase)
```

```python
df.head()
```

### Remove HTML Tags

```python
import re 
def remove_html_tags(text): 
    pattern = re.compile("<.*?>")
    return pattern.sub(r'' , text)
```

```python
df['review'] = df['review'].apply(remove_html_tags)
```

```python
df.head()
```

### Remove Punctuations

```python
import string
df['review'] = df['review'].str.replace(
    f"[{string.punctuation}]", "", regex = True 
)
```

```python
punc_list = string.punctuation
punc_list
```

#### Efficient Way to remove punctuations

```python
def remove_punctuations(text): 
    return text.translate(str.maketrans('' , '',punc_list))
```

- Note: maketrans(Characters to replace, Characters to replace them with, Characters to delete)
    - '': no character to replace
    - '': no character to replace with
    - punc_list: characters to delete from the text

### Stop words removing

```python
from nltk.corpus import stopwords
```

```python
import nltk
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
stopwords.words('bengali')
```

```python
stop_words = set(stopwords.words('english'))
def remove_stop_words(text): 
    return " ".join(word for word in text.split() if word not in stop_words)
```

```python
df['review'] = df['review'].apply(remove_stop_words)
```

```python
ps = PorterStemmer()
def stem_words(text): 
    return " ".join([ps.stem(word) for word in text.split()])
```
