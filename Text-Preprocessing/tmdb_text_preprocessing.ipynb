```python
# check is there html tags present?
import re
pattern = re.compile("<.*?>")
def has_html_tags(text): 
    return bool(pattern.search(text))
```

```python
df['description'].isnull().sum()
```

```python
df.dropna(inplace = True)
```

```python
df['description'].apply(has_html_tags).sum()
```

- No HTML tags are present

### Remove Punctuations and do lower casing

```python
import string
punc_list = string.punctuation
def remove_punctuations(text): 
    text = text.lower()
    return text.translate(str.maketrans("" , "" , punc_list))
```

```python
temp = "HI! My name is tipto. My age is 23. I have interest in ml,dl,nlp,llms."
remove_punctuations(temp)
```

```python
df['description'] = df['description'].apply(remove_punctuations)
```

### Remove stop words

```python
from nltk.corpus import stopwords
```

```python
import nltk
nltk.download('stopwords')
```

```python
eng_stopwords = set(stopwords.words('english'))
def remove_stop_words(text): 
    return " ".join([word for word in text.split() if word not in eng_stopwords])
```

```python
df['description'] = df['description'].apply(remove_stop_words)
```
