- download the word2vec model bin from [here](https://www.kaggle.com/datasets/suraj520/googlenews-vectors-negative300bingz-gz-format) and extract it to use.

```python
word2vec_path = "embeddings/word2vec/GoogleNews-vectors-negative300.bin"
w2v = KeyedVectors.load_word2vec_format(word2vec_path , binary = True)
print("Loaded vocabulary size:", len(w2v))
```

```python
w2v['man'].shape
```

```python
w2v.most_similar('man')
```

```python
w2v.most_similar('football')
```

```python
# Find similiarity between 2 words
w2v.similarity('cricket' , 'football')
```

```python
w2v.similarity('man' , 'coding')
```

```python
vec = w2v['king'] - w2v['man'] + w2v['woman'] 
w2v.most_similar([vec])
```
